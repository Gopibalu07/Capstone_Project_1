Chart 1 — NAV Trend Analysis

```python
nav["date"] = pd.to_datetime(nav["date"])

plt.figure(figsize=(15,7))

for code in nav["amfi_code"].unique():
    temp = nav[nav["amfi_code"]==code]
    plt.plot(temp["date"],temp["nav"])

plt.title("NAV Trend")
plt.savefig("../reports/charts/01_nav_trend.png",
            dpi=300,
            bbox_inches="tight")
plt.show()
```

Chart 2 — AUM Growth

```python
plt.figure(figsize=(12,6))

sns.barplot(
    data=aum,
    x="fund_house",
    y="aum_crore",
    hue="date"
)

plt.xticks(rotation=90)

plt.savefig("../reports/charts/02_aum_growth.png",
            dpi=300,
            bbox_inches="tight")

plt.show()
```

Chart 3 — SIP Trend

```python
plt.figure(figsize=(12,6))

plt.plot(
    sip["month"],
    sip["sip_inflow_crore"],
    marker="o"
)

plt.title("Monthly SIP Inflow")
plt.xlabel("Month")
plt.ylabel("SIP Inflow (Crores)")

plt.xticks(rotation=45)
plt.grid(True)

plt.tight_layout()

plt.savefig(
    "../reports/charts/03_sip_trend.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

Chart 4 — Category Heatmap

```python
pivot = category.pivot_table(
    index="category",
    columns="month",
    values="net_inflow_crore"
)

plt.figure(figsize=(14,6))

sns.heatmap(
    pivot,
    cmap="YlGnBu"
)

plt.savefig("../reports/charts/04_category_heatmap.png",
            dpi=300,
            bbox_inches="tight")

plt.show()
```

Chart 5 — Age Group Distribution

```python
transactions["age_group"].value_counts().plot.pie(
    autopct="%1.1f%%"
)
plt.savefig("../reports/charts/05_age_distribution.png",
            dpi=300,
            bbox_inches="tight")
plt.show()
```

Chart 6 — SIP Amount by Age

```python
sns.boxplot(
    data=transactions,
    x="age_group",
    y="amount_inr"
)
plt.savefig("../reports/charts/06_sip_amount_boxplot.png",
            dpi=300,
            bbox_inches="tight")
plt.show()
```

Chart 7 — State-wise Investment

```python
state = transactions.groupby(
    "state"
)["amount_inr"].sum()

state.sort_values().plot.barh(
    figsize=(12,8)
)
plt.savefig("../reports/charts/07_state_investment.png",
            dpi=300,
            bbox_inches="tight")
plt.show()
```

Chart 8 — T30 vs B30

```python
transactions["city_tier"].value_counts().plot.pie(
    autopct="%1.1f%%"
)
plt.savefig("../reports/charts/08_city_tier_distribution.png",
            dpi=300,
            bbox_inches="tight")
plt.show()
```

Chart 9 — Folio Growth

```python
folio["month"] = pd.to_datetime(folio["month"])

plt.figure(figsize=(12,6))

plt.plot(
    folio["month"],
    folio["total_folios_crore"],
    marker="o",
    linewidth=2
)

plt.title("Folio Growth Over Time")
plt.xlabel("Month")
plt.ylabel("Total Folios (Crore)")

plt.xticks(rotation=45)
plt.grid(True)

plt.tight_layout()

plt.savefig(
    "../reports/charts/09_folio_growth.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

plt.close()
```

Chart 10 — Correlation Matrix

```python
pivot = nav.pivot_table(
    index="date",
    columns="amfi_code",
    values="nav"
)

corr = pivot.corr()

plt.figure(figsize=(12,10))

sns.heatmap(
    corr,
    cmap="coolwarm"
)
plt.savefig("../reports/charts/10_correlation_heatmap.png",
            dpi=300,
            bbox_inches="tight")
plt.show()
```

Expense Ratio Distribution (Histogram)

Fund Category Count (Bar Chart)

```python
plt.figure(figsize=(12,6))

fund["category"].value_counts().plot(
    kind="bar",
    color="steelblue"
)

plt.title("Number of Funds by Category")
plt.xlabel("Category")
plt.ylabel("Count")

plt.xticks(rotation=45)
plt.savefig("../reports/charts/13_fund_category_count.png",
            dpi=300,
            bbox_inches="tight")
plt.show()
```

Monthly Transaction Volume (Line Chart)

```python
transactions["transaction_date"] = pd.to_datetime(
    transactions["transaction_date"]
)
```

```python
monthly = (
    transactions
    .groupby(
        transactions["transaction_date"].dt.to_period("M")
    )["amount_inr"]
    .sum()
)

monthly.index = monthly.index.astype(str)

plt.figure(figsize=(14,6))

plt.plot(
    monthly.index,
    monthly.values,
    marker="o",
    linewidth=2
)

plt.title("Monthly Transaction Volume")
plt.xlabel("Month")
plt.ylabel("Total Transaction Amount (INR)")

plt.xticks(rotation=45)

plt.grid(True)
plt.savefig("../reports/charts/14_monthly_transaction_volume.png",
            dpi=300,
            bbox_inches="tight")
plt.show()
```

Transaction Type Distribution (Donut Chart)

```python
transaction_counts = (
    transactions["transaction_type"]
    .value_counts()
)

plt.figure(figsize=(7,7))

plt.pie(
    transaction_counts,
    labels=transaction_counts.index,
    autopct="%1.1f%%",
    startangle=90,
    wedgeprops={"width":0.4}
)

plt.title("Transaction Type Distribution")
plt.savefig("../reports/charts/15_transaction_type_distribution.png",
            dpi=300,
            bbox_inches="tight")
plt.show()
```

Gender-wise Investment (Bar Chart)

```python
gender = (
    transactions
    .groupby("gender")["amount_inr"]
    .sum()
)

plt.figure(figsize=(8,5))

sns.barplot(
    x=gender.index,
    y=gender.values
)

plt.title("Gender-wise Investment")
plt.xlabel("Gender")
plt.ylabel("Investment Amount (INR)")
plt.savefig("../reports/charts/16_gender_wise_investment.png",
            dpi=300,
            bbox_inches="tight")
plt.show()
```

# EDA Findings

1. NAV values show stable long-term growth across most mutual fund schemes.

2. Major fund houses continue to dominate the industry's Assets Under Management (AUM).

3. Monthly SIP inflows have increased steadily, indicating growing retail investor confidence.

4. Equity funds attract the highest net inflows among all fund categories.

5. Investors aged 26–35 represent the largest investor segment.

6. Investment amounts vary widely, with a few high-value transactions acting as outliers.

7. State-wise analysis shows that a small number of states contribute the majority of investments.

8. T30 cities generate significantly higher investments than B30 cities.

9. Total folio count has grown consistently throughout the analysis period.

10. Strong positive correlations exist between several mutual fund schemes, indicating similar market movement.

11. Financial Services and Information Technology are the most heavily weighted portfolio sectors.

12. Most mutual fund schemes have expense ratios within acceptable industry standards.

13. Equity schemes account for the highest number of mutual fund offerings.

14. SIP is the most preferred transaction type, followed by Lumpsum and Redemption.

15. Gender-wise investment analysis shows noticeable differences in investment participation between male and female investors.
